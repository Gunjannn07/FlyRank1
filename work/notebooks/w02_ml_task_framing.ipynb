## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

The pattern becomes messy because a page's CTR opportunity depends on more than one factor. For example: a page can have low CTR but also low search volume, or have a good CTR while ranking poorly. A fixed rule would only look at one or two conditions. Whereas, ML can look at several factors together and find patterns that a fixed rule might miss.
